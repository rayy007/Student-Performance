```python
from sklearn.model_selection import GridSearchCV

def optimize_models(models, param_grids, X_train, y_train):
    best_models = {}

    for name, model in models.items():
        print(f"Optimizing {name}...")

        grid = GridSearchCV(
            estimator=model,
            param_grid=param_grids[name],
            cv=5,
            scoring="accuracy",
            n_jobs=-1
        )

        grid.fit(X_train, y_train)

        best_models[name] = grid.best_estimator_

        print("Best Parameters:", grid.best_params_)
        print("Best Score:", grid.best_score_)
        print("-" * 50)

    return best_models
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
from xgboost import XGBClassifier
from lightgbm import LGBMClassifier
from catboost import CatBoostClassifier

# Logistic Regression
def logistic_model():
    return LogisticRegression(random_state=42, max_iter=1000)

# Decision Tree
def decision_tree_model():
    return DecisionTreeClassifier(random_state=42)

# Random Forest
def random_forest_model():
    return RandomForestClassifier(random_state=42)

# SVM
def svm_model():
    return SVC(probability=True, random_state=42)

# XGBoost
def xgb_model():
    return XGBClassifier(
        random_state=42,
        eval_metric='logloss'
    )

# LightGBM
def lgbm_model():
    return LGBMClassifier(random_state=42)

# CatBoost
def catboost_model():
    return CatBoostClassifier(
        random_state=42,
        verbose=0
    )
```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
import pandas as pd

def evaluate_models(models, X_train, X_test, y_train, y_test):
    results = []

    for name, model in models.items():
        model.fit(X_train, y_train)

        # Train prediction
        y_train_pred = model.predict(X_train)

        # Test prediction
        y_test_pred = model.predict(X_test)

        results.append({
            "Model": name,
            "Train Accuracy": accuracy_score(y_train, y_train_pred),
            "Test Accuracy": accuracy_score(y_test, y_test_pred),
            "Train Precision": precision_score(y_train, y_train_pred, average='weighted'),
            "Test Precision": precision_score(y_test, y_test_pred, average='weighted'),
            "Train Recall": recall_score(y_train, y_train_pred, average='weighted'),
            "Test Recall": recall_score(y_test, y_test_pred, average='weighted'),
            "Train F1 Score": f1_score(y_train, y_train_pred, average='weighted'),
            "Test F1 Score": f1_score(y_test, y_test_pred, average='weighted')
        })

    return pd.DataFrame(results)
```

```python
models = {
    "Logistic Regression": logistic_model(),
    "Decision Tree": decision_tree_model(),
    "Random Forest": random_forest_model(),
    "SVM": svm_model(),
    "XGBoost": xgb_model(),
    "LightGBM": lgbm_model(),
    "CatBoost": catboost_model()
}

results_df = evaluate_models(models, X_train, X_test, y_train, y_test)

results_df
```
